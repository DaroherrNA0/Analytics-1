Descripción: Se hizo una busqueda de empresas sobre 5000 trabajadores con IA solicitando el nombre y rut de estas. Posterior a ese se requiere saber cuales de estas ya pertenecen a MUTUAL.
Este SCRIPT permite hacer un nuevo documento excel que permite saber cuales empresas ya estan adheridas a mutual, cuales no y cuales estuviero en algun momento.

```python
from pathlib import Path
import pandas as pd

# 1) Leer un archivo Excel con ADHERENTES exacta
Posibles_ad = pd.read_excel(r"C:\Users\drojas\OneDrive - Mutual\Escritorio\Posibles adherentes.xlsx")
ADHERENTES  = pd.read_excel(r"C:\Users\drojas\OneDrive - Mutual\Escritorio\ADHERENTES MUTUAL.xlsx")
```

```python
ADHERENTES.columns
```

```python
import pandas as pd

# Posibles_ad: tiene columna "RUT"
# ADHERENTES:  tiene columna "ADHR_RUT_EMPRESA"


base_adherentes = (
    ADHERENTES[["ADHR_RUT_EMPRESA", "ADHR_ESTADO"]]
    .drop_duplicates(subset=["ADHR_RUT_EMPRESA"])
)

out = Posibles_ad.merge(
    base_adherentes,
    left_on="RUT",
    right_on="ADHR_RUT_EMPRESA",
    how="left"
)

#Columna "Mutual?" (Si si hubo match, No si no)
out["Mutual?"] = out["ADHR_RUT_EMPRESA"].notna().map({True: "Si", False: "No"})


out.to_excel(r"C:\Users\drojas\OneDrive - Mutual\Escritorio\MATCH.xlsx", index=False, engine="openpyxl")
```
